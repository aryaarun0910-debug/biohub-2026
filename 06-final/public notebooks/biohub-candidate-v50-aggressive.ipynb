```python
# ==============================================================================
# Cell 2: Master SOTA Post-Link Layer (Deterministic Linearization + Cytokinesis SOTA)
# ==============================================================================
from collections import Counter, defaultdict
import json
from pathlib import Path
import numpy as np
import pandas as pd
from scipy.spatial import cKDTree

SCALE = np.asarray((1.625, 0.40625, 0.40625), dtype=np.float64)
COLUMNS = ["id", "dataset", "row_type", "node_id", "t", "z", "y", "x", "source_id", "target_id"]

v1329_path = Path("/kaggle/working/v1329_submission.csv")
final_path = Path("/kaggle/working/submission.csv")

assert v1329_path.is_file(), f"Missing {v1329_path}"

v1329_frame = pd.read_csv(v1329_path)
print(f"V1329 foundation rows: {len(v1329_frame):,}")

def pos_um(coords, nid):
    c = coords[nid]
    return np.asarray((c[1], c[2], c[3]), dtype=np.float64) * SCALE

def edge_dist_um(c1, c2):
    return float(np.linalg.norm((np.asarray(c1[1:], dtype=np.float64) - np.asarray(c2[1:], dtype=np.float64)) * SCALE))

datasets = sorted(v1329_frame["dataset"].unique())
print(f"Processing {len(datasets)} datasets dynamically: {datasets}...")

final_rows = []
audit_records = []

for ds in datasets:
    part = v1329_frame[v1329_frame["dataset"] == ds]
    nodes = part[part["row_type"] == "node"]
    edges = part[part["row_type"] == "edge"]

    node_coords = {
        int(r.node_id): (int(r.t), float(r.z), float(r.y), float(r.x))
        for r in nodes.itertuples()
    }
    raw_edges = [(int(r.source_id), int(r.target_id)) for r in edges.itertuples()]

    # Step 1: Linearize raw V1329 forks to purge 100% of division false positives
    out_edges = defaultdict(list)
    for src, tgt in raw_edges:
        out_edges[src].append(tgt)

    lin_edges = []
    dropped_forks = 0
    for src, tgts in out_edges.items():
        if len(tgts) == 1:
            lin_edges.append((src, tgts[0]))
        elif len(tgts) == 2:
            p_pos = pos_um(node_coords, src)
            c1_pos = pos_um(node_coords, tgts[0])
            c2_pos = pos_um(node_coords, tgts[1])
            d1 = float(np.linalg.norm(p_pos - c1_pos))
            d2 = float(np.linalg.norm(p_pos - c2_pos))
            chosen = tgts[0] if d1 <= d2 else tgts[1]
            lin_edges.append((src, chosen))
            dropped_forks += 1
        else:
            for t in tgts:
                lin_edges.append((src, t))

    # Step 2: Physical Cytokinesis Recovery in dense films (len(nodes) >= 30,000)
    final_edges = list(lin_edges)
    added_cytokinesis = []
    if len(node_coords) >= 30000:
        in_deg = defaultdict(int)
        out_edges_lin = defaultdict(list)
        for src, tgt in lin_edges:
            out_edges_lin[src].append(tgt)
            in_deg[tgt] += 1

        def track_len(nid):
            curr = nid
            l = 1
            while curr in out_edges_lin and len(out_edges_lin[curr]) == 1:
                curr = out_edges_lin[curr][0]
                l += 1
            return l

        times = defaultdict(list)
        for nid, c in node_coords.items():
            times[c[0]].append(nid)

        proposals = []
        for t in sorted(times):
            child_frame_ids = times.get(t + 1, [])
            if not child_frame_ids:
                continue
            source_ids = [nid for nid in times[t] if len(out_edges_lin.get(nid, [])) == 1]
            cand_ids = [nid for nid in child_frame_ids if in_deg[nid] == 0]
            if not source_ids or not cand_ids:
                continue

            cand_positions = np.stack([pos_um(node_coords, cid) for cid in cand_ids])
            tree = cKDTree(cand_positions)

            for s in source_ids:
                c1 = out_edges_lin[s][0]
                child_dist = edge_dist_um(node_coords[s], node_coords[c1])
                if child_dist > 8.5:
                    continue
                _, nn_idx = tree.query(pos_um(node_coords, c1))
                mutual_nn_id = cand_ids[int(nn_idx)]

                for c2 in cand_ids:
                    if c2 != mutual_nn_id:
                        continue
                    parent_dist = edge_dist_um(node_coords[s], node_coords[c2])
                    if parent_dist > 8.5:
                        continue
                    sister_dist = edge_dist_um(node_coords[c1], node_coords[c2])
                    if sister_dist > 13.5 or sister_dist < 8.5:
                        continue

                    coords_p = node_coords[s]
                    coords_c1 = node_coords[c1]
                    coords_c2 = node_coords[c2]
                    if min(coords_p[3], coords_c1[3], coords_c2[3]) < 20.0 or max(coords_p[3], coords_c1[3], coords_c2[3]) > 236.0:
                        continue
                    if min(coords_p[2], coords_c1[2], coords_c2[2]) < 20.0 or max(coords_p[2], coords_c1[2], coords_c2[2]) > 236.0:
                        continue

                    l1 = track_len(c1)
                    l2 = track_len(c2)
                    if min(l1, l2) < 12:
                        continue

                    c1_succ = out_edges_lin.get(c1, [])
                    c2_succ = out_edges_lin.get(c2, [])
                    if len(c1_succ) != 1 or len(c2_succ) != 1:
                        continue
                    gc1 = c1_succ[0]
                    gc2 = c2_succ[0]
                    gc_dist = edge_dist_um(node_coords[gc1], node_coords[gc2])
                    div = gc_dist - sister_dist
                    if div < 1.20:
                        continue

                    v1 = pos_um(node_coords, c1) - pos_um(node_coords, s)
                    v2 = pos_um(node_coords, c2) - pos_um(node_coords, s)
                    cos_ang = np.dot(v1, v2) / (np.linalg.norm(v1) * np.linalg.norm(v2) + 1e-6)
                    ang_deg = float(np.degrees(np.arccos(np.clip(cos_ang, -1.0, 1.0))))
                    if ang_deg < 140.0:
                        continue

                    com = (pos_um(node_coords, c1) + pos_um(node_coords, c2)) / 2.0
                    com_dev = float(np.linalg.norm(pos_um(node_coords, s) - com))
                    if com_dev > 3.2:
                        continue

                    c1_pos = pos_um(node_coords, c1)
                    c2_pos = pos_um(node_coords, c2)
                    dz_s = abs(c1_pos[0] - c2_pos[0])
                    dxy_s = float(np.sqrt((c1_pos[1] - c2_pos[1])**2 + (c1_pos[2] - c2_pos[2])**2))
                    if dxy_s < 9.6 or dz_s > 4.0:
                        continue

                    persistence_bonus = min(l1, l2) / 30.0
                    fitness = (ang_deg / 180.0) + (1.0 - com_dev/2.8) + (div / 5.0) + persistence_bonus
                    proposals.append((fitness, s, c2, ang_deg, com_dev, div, dxy_s, dz_s, t))

        proposals.sort(key=lambda x: -x[0])
        used_s = set()
        used_c = set()
        for fitness, s, c2, ang, com_dev, div, dxy_s, dz_s, t in proposals:
            if s not in used_s and c2 not in used_c and len(added_cytokinesis) < 2:
                used_s.add(s)
                used_c.add(c2)
                added_cytokinesis.append((s, c2))
                final_edges.append((s, c2))
                print(f"[{ds}] Rescued cytokinesis division: {s} -> {c2} (fitness={fitness:.3f}, t={t})")

    # Step 3: Enforce strict DAG & competition invariants
    out_counts = Counter(src for src, _ in final_edges)
    in_counts = Counter(tgt for _, tgt in final_edges)

    assert max(out_counts.values(), default=0) <= 2, f"Out-degree violation in {ds}"
    assert max(in_counts.values(), default=0) <= 1, f"In-degree violation in {ds}"
    assert all(src in node_coords and tgt in node_coords for src, tgt in final_edges), f"Edge references missing node in {ds}"
    assert all(node_coords[tgt][0] == node_coords[src][0] + 1 for src, tgt in final_edges), f"Temporal edge violation in {ds}"

    # Build rows
    for nid, c in sorted(node_coords.items()):
        final_rows.append({
            "id": None, "dataset": ds, "row_type": "node",
            "node_id": nid, "t": c[0], "z": c[1], "y": c[2], "x": c[3],
            "source_id": -1, "target_id": -1,
        })
    for src, tgt in sorted(final_edges):
        final_rows.append({
            "id": None, "dataset": ds, "row_type": "edge",
            "node_id": -1, "t": -1, "z": -1.0, "y": -1.0, "x": -1.0,
            "source_id": src, "target_id": tgt,
        })

    audit_records.append({
        "dataset": ds, "nodes": len(node_coords), "edges": len(final_edges),
        "dropped_raw_forks": dropped_forks, "rescued_cytokinesis": len(added_cytokinesis),
    })

# Format output dataframe
output_df = pd.DataFrame(final_rows)
output_df["id"] = np.arange(len(output_df), dtype=np.int64)
output_df = output_df[COLUMNS]

output_df.to_csv(final_path, index=False)
print(f"V50 Master submission successfully saved to {final_path}!")
print(f"Total rows: {len(output_df):,}")
print("Dataset summary:")
for r in audit_records:
    print(f"  {r['dataset']}: nodes={r['nodes']:,}, edges={r['edges']:,}, dropped_forks={r['dropped_raw_forks']}, cytokinesis_rescued={r['rescued_cytokinesis']}")
```
